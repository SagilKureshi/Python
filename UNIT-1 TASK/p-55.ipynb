```python
import tkinter as tk

# Create main window
window = tk.Tk()
window.title("Simple GUI")
window.geometry("400x250")

# Label
label = tk.Label(window, text="Enter your name:")
label.pack(pady=10)

# Entry field
entry = tk.Entry(window)
entry.pack(pady=5)

# Function for button
def show_message():
    name = entry.get()
    result.config(text="Hello " + name)

# Button
button = tk.Button(window, text="Submit", command=show_message)
button.pack(pady=10)

# Result label
result = tk.Label(window, text="")
result.pack(pady=10)

# Run GUI
window.mainloop()
```
