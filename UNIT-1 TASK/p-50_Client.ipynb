TCP/IP Client

```python
import socket

# Create client socket
client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

# Connect to server
client.connect(("127.0.0.1", 5000))

# Send message
client.send("Hello from Client".encode())

# Receive response
message = client.recv(1024).decode()
print("Server message:", message)

# Close connection
client.close()
```
