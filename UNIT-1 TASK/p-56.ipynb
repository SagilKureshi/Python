```python
import tkinter as tk
from tkinter import messagebox

# Create main window
window = tk.Tk()
window.title("Dialog Box Example")
window.geometry("400x250")


# Function for information dialog
def show_info():
    messagebox.showinfo("Information", "This is an information message.")


# Function for warning dialog
def show_warning():
    messagebox.showwarning("Warning", "This is a warning message.")


# Function for confirmation dialog
def show_confirmation():
    answer = messagebox.askyesno(
        "Confirmation",
        "Do you want to continue?"
    )

    if answer:
        messagebox.showinfo("Result", "You selected Yes.")
    else:
        messagebox.showinfo("Result", "You selected No.")


# Buttons
tk.Button(
    window,
    text="Information",
    command=show_info
).pack(pady=10)

tk.Button(
    window,
    text="Warning",
    command=show_warning
).pack(pady=10)

tk.Button(
    window,
    text="Confirmation",
    command=show_confirmation
).pack(pady=10)


# Run GUI
window.mainloop()
```
