```python
import pylab as plt

# Data
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
```

```python
# 1. Line Plot
plt.figure()
plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# 2. Bar Chart
plt.figure()
plt.bar(x, y)
plt.title("Bar Chart")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# 3. Pie Chart
plt.figure()
values = [30, 20, 25, 25]
labels = ["A", "B", "C", "D"]

plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Pie Chart")
plt.show()
```

```python
# 4. Histogram
plt.figure()
data = [10, 12, 15, 15, 18, 20, 20, 22, 25, 25, 28, 30]

plt.hist(data, bins=5)
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
# 5. Scatter Plot
plt.figure()
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```
