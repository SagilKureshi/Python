UDP Client

```python
import socket

# Create UDP socket
client = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)

# Send message to server
client.sendto("Hello from UDP Client".encode(), ("127.0.0.1", 5002))

# Receive response
message, address = client.recvfrom(1024)

print("Server:", message.decode())

client.close()
```
