UDP Server

```python
import socket

# Create UDP socket
server = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)

# Bind server
server.bind(("127.0.0.1", 5002))

print("UDP Server is waiting for message...")

# Receive message
message, address = server.recvfrom(1024)

print("Client:", message.decode())

# Send response
server.sendto("Hello from UDP Server".encode(), address)

server.close()

print("Server closed.")
```
