```python
import tkinter as tk

# Create main window
window = tk.Tk()
window.title("Simple Calculator")
window.geometry("300x400")

# Entry field
entry = tk.Entry(window, width=20, font=("Arial", 20), justify="right")
entry.grid(row=0, column=0, columnspan=4, padx=10, pady=20)


# Function to add value to entry
def click(value):
    entry.insert(tk.END, value)


# Function to calculate result
def calculate():
    try:
        result = eval(entry.get())
        entry.delete(0, tk.END)
        entry.insert(0, result)
    except:
        entry.delete(0, tk.END)
        entry.insert(0, "Error")


# Function to clear entry
def clear():
    entry.delete(0, tk.END)


# Buttons
buttons = [
    ("7", 1, 0), ("8", 1, 1), ("9", 1, 2), ("/", 1, 3),
    ("4", 2, 0), ("5", 2, 1), ("6", 2, 2), ("*", 2, 3),
    ("1", 3, 0), ("2", 3, 1), ("3", 3, 2), ("-", 3, 3),
    ("0", 4, 0), (".", 4, 1), ("+", 4, 2)
]

for text, row, column in buttons:
    tk.Button(
        window,
        text=text,
        width=5,
        height=2,
        font=("Arial", 14),
        command=lambda value=text: click(value)
    ).grid(row=row, column=column, padx=5, pady=5)

# Equal button
tk.Button(
    window,
    text="=",
    width=5,
    height=2,
    font=("Arial", 14),
    command=calculate
).grid(row=4, column=3, padx=5, pady=5)

# Clear button
tk.Button(
    window,
    text="Clear",
    width=23,
    height=2,
    command=clear
).grid(row=5, column=0, columnspan=4, padx=5, pady=10)

# Run GUI
window.mainloop()
```
