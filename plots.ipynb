```python
import matplotlib.pyplot as plt
import pylab as pl
```

```python
x =[1,2,3,4,5]
y = [10,20,30,40,50]
plt.plot(x,y)
pl.show()
```

Property of plot

```python
x =[1,2,3,4,5]
y = [10,5,15,20,30]
plt.plot(x,y,marker='D',linestyle='--',linewidth='5',ms='20',mfc='r',mec='yellowgreen',mew='3')
pl.show()
```

# xlim() and ylim()

```python
x =[1,2,3,4,5]
y = [10,5,10,5,10]
plt.plot(x,y)
plt.xlim(0,10)
plt.ylim(0,20)
pl.show()
```

# Many plot


```python
x1 =[1,2,3,4,5]
x2 =[1,4,3,2,1]
y1 = [10,5,10,5,10]
y2 = [0,5,0,5,0]
plt.plot(x1,y1)
plt.plot(x2,y2)
plt.xlim(0,10)
plt.ylim(0,20)
pl.show()
```

# Subplot

```python
x = [0,1,2,3]
y = [3,8,1,10]

plt.subplot(2,1,1)
plt.title("plote 1")
plt.plot(x,y)

x = [0,1,2,3]
y = [10,20,30,40]

plt.subplot(2,1,2)
plt.title("plote 2")
plt.plot(x,y)
plt.suptitle("Main title")
plt.show()
```
