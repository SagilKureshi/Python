```python
import matplotlib.pyplot as plt
import numpy as np
```

# All Chart

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]

plt.subplot(2,3,1)
plt.plot(x,y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Plote 1")


plt.subplot(2,3,2)
plt.bar(x,y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Plote 2")

plt.subplot(2,3,3)
plt.scatter(x,y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Plote 3")

plt.subplot(2,3,4)
plt.barh(x,y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Plote 4")

plt.subplot(2,3,5)
plt.pie(x)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Plote 5")

x2 = np.random.normal(170,10,250)
plt.subplot(2,3,6)
plt.hist(x2)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Plote 6")
plt.tight_layout()
```
